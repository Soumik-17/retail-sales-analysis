```python
import pandas as pd
```

```python
file_09_10 = r"C:\Users\soumi\Desktop\online_retail_09_10.csv"
file_10_11 = r"C:\Users\soumi\Desktop\online_retail_10_11.csv"

output_file = r"C:\Users\soumi\Desktop\online_retail_combined.csv"
```

```python
df_09_10 = pd.read_csv(
    file_09_10,
    encoding="utf-8-sig"
)

df_10_11 = pd.read_csv(
    file_10_11,
    encoding="utf-8-sig"
)
```

```python
df_09_10.columns = df_09_10.columns.str.strip()
df_10_11.columns = df_10_11.columns.str.strip()
```

```python
df = pd.concat(
    [df_09_10, df_10_11],
    ignore_index=True
)
```

```python
df = df.dropna(how="all")
```

```python
text_columns = [
    "InvoiceNo",
    "StockCode",
    "Description",
    "Country"
]

for col in text_columns:
    df[col] = df[col].astype("string").str.strip()
```

```python
df["CustomerID"] = (
    df["CustomerID"]
    .astype("string")
    .str.strip()
)
```

```python
df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)

df["InvoiceDate"] = df["InvoiceDate"].dt.strftime(
    "%Y-%m-%d %H:%M:%S"
)
```

```python
df["Quantity"] = pd.to_numeric(
    df["Quantity"],
    errors="coerce"
)

df["UnitPrice"] = pd.to_numeric(
    df["UnitPrice"],
    errors="coerce"
)
```

```python
df = df.dropna(
    subset=[
        "InvoiceNo",
        "StockCode",
        "Quantity",
        "InvoiceDate",
        "UnitPrice",
        "Country"
    ]
)
```

```python
df = df.drop_duplicates()
```

```python
df = df.reset_index(drop=True)
```

```python
df.to_csv(
    output_file,
    index=False,
    encoding="utf-8",
    quoting=1   # csv.QUOTE_ALL
)
```

```python
print("Combined dataset created successfully.")
print()
print("Rows:", len(df))
print("Columns:", len(df.columns))
print()
print("Columns:")
print(df.columns.tolist())
print()
print("Saved to:")
print(output_file)
```
